## Prespecified decisions

- Random seed: 42.
- Bootstrap resamples: 2,000.
- Stable satellite subset: accepted exact-original matches whose normalized district name is globally unique in both the 709 complete DFSI records and the 735 ADM2 geometries.
- Higher-reporting-quality subset: at least one matched event in 1986-2004 and fatality-field completeness of at least 50% in both 1986-2004 and 2005-2023. Thresholds of 25% and 75% are sensitivity checks only.
- Fractional fatalities: event fatalities divided by the total number of listed districts; only matched listed shares are assigned, and unmatched shares remain unassigned.
